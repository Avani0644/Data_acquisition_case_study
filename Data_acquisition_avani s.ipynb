```python
#TASK 6

# Excel file
excel_data = pd.read_excel('gdp.xlsx')

# JSON file
json_data = pd.read_json('internet_users.json')

# XML file
literacy_df = pd.read_xml('literacy_rate.xml')

# CSV file
population_df = pd.read_csv('population.csv')
# Function to clean column names
def clean_columns(df):
    df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
    return df

# Apply cleaning
excel_data = clean_columns(excel_data)
json_data = clean_columns(json_data)
literacy_df = clean_columns(literacy_df)
population_df = clean_columns(population_df)

# Rename to common key column
literacy_df.rename(columns={'name': 'country'}, inplace=True)

# Display column names
print(excel_data.columns)
print(json_data.columns)
print(literacy_df.columns)
print(population_df.columns)
```

```python
#TASK -7
excel_data['gdp'] = pd.to_numeric(excel_data['gdp'])
json_data['internet_users'] = pd.to_numeric(json_data['internet_users'])
literacy_df['literacy_rate'] = pd.to_numeric(literacy_df['literacy_rate'])
population_df['population'] = pd.to_numeric(population_df['population'])

print("Excel Data:\n", excel_data.dtypes)
print("JSON Data:\n", json_data.dtypes)
print("Literacy Data:\n", literacy_df.dtypes)
print("Population Data:\n", population_df.dtypes)
```

```python
#TASK 8
merged_df = pd.merge(excel_data, json_data)
merged_df = pd.merge(merged_df, literacy_df)
merged_df = pd.merge(merged_df, population_df)

print(merged_df.head())
```

```python
#TASK 9
# Internet Penetration (%)
merged_df['internet_penetration'] = (merged_df['internet_users'] / merged_df['population']) * 100
merged_df['internet_penetration'] = merged_df['internet_penetration'].round(2)

# GDP Per Capita
merged_df['gdp_per_capita'] = merged_df['gdp'] / merged_df['population']
merged_df['gdp_per_capita'] = merged_df['gdp_per_capita'].round(2)

# Show output
print(merged_df.head())
```

```python
#TASK 10
import sqlite3

# Create database (or connect if already exists)
conn = sqlite3.connect('country_data.db')

# Store DataFrame into table
merged_df.to_sql('country_statistics', conn, if_exists='replace', index=False)

# Close connection
conn.close()

print("Data stored successfully!")
```

ETL Process Summary

Data Sources Used:
Data was collected from multiple sources including Excel (GDP data), JSON (internet users), and CSV files (literacy rate and population).

Data Cleaning Steps:
Column names were standardized (converted to lowercase, removed spaces), and renamed to ensure a common key column (country) across all datasets.

Transformations Performed:
Data types were converted to numeric format, and new columns such as Internet Penetration (%) and GDP Per Capita were calculated and rounded.

Data Merging:
All datasets were merged into a single DataFrame using the common column country.

Database Loading Process:
The final dataset was stored in a SQLite database (country_data.db) in a table named country_statistics.

Key Insights Obtained:
Analysis revealed top countries by GDP and population, average literacy and internet usage, and a positive correlation between literacy rate and internet penetration
